```python
import pandas as pd
import matplotlib.pyplot as plt

print("Gene Expression Analysis Project")
print("Libraries imported successfully!")
```

```python
data = {
    'Gene': ['BRCA1', 'TP53', 'EGFR', 'MYC', 'KRAS',
             'BRAF', 'APC', 'PTEN', 'AKT1', 'CDK2',
             'VEGFA', 'ERBB2', 'GAPDH', 'ACTB', 'BCL2'],

    'Expression': [450, 890, 670, 950, 720,
                   380, 560, 490, 810, 610,
                   920, 750, 300, 400, 530]
}

df = pd.DataFrame(data)

df.to_csv('sample_gene_expression.csv', index=False)

print("CSV file created successfully!")
print(df)
```

```python
# Sort genes by expression value
sorted_genes = df.sort_values(
    by='Expression',
    ascending=False
)

# Select the top 10 genes
top_10_genes = sorted_genes.head(10)

# Display the result
print("Top 10 Highly Expressed Genes")
print(top_10_genes)
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    top_10_genes['Gene'],
    top_10_genes['Expression']
)

plt.xlabel("Gene Names")
plt.ylabel("Expression Level")
plt.title("Top 10 Highly Expressed Genes")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Save top 10 genes in CSV format
top_10_genes.to_csv("top_10_genes.csv", index=False)

# Save graph as an image
plt.figure(figsize=(10, 6))

plt.bar(top_10_genes['Gene'],
        top_10_genes['Expression'])

plt.xlabel("Gene Names")
plt.ylabel("Expression Level")
plt.title("Top 10 Highly Expressed Genes")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("top_10_genes.png", dpi=300)

plt.show()

print("Both files saved successfully!")
```
